```python
orders
```

```python
ml_data = orders[orders['shipping status'] != 'not shipped']
ml_data 
ml_data['shipping status'].value_counts()
ml_data['shipping status'] = ml_data['shipping status'].map({'on time':0,'late':1})
ml_data['shipping status'].value_counts()
```

```python
ml_data['year'] = ml_data['orderDate'].dt.year
ml_data['month'] = ml_data['orderDate'].dt.month
ml_data['day_of_week'] = ml_data['orderDate'].dt.dayofweek
ml_data['allowed_days']=(ml_data['requiredDate']-ml_data['orderDate']).dt.days
```

```python
ml_data
```

```python
order_details
```

```python
Total_quantity_per_order=order_details.groupby('orderID')['quantity'].sum().reset_index(name='total_quantity_per_order')
Total_quantity_per_order
ml_data=ml_data.merge(Total_quantity_per_order,on='orderID')
```

```python
ml_data
```

```python
total_sales_per_order
```

```python
ml_data = ml_data.drop(columns=['sales', 'sales_x', 'sales_y'], errors='ignore')
ml_data
```

```python
ml_data = ml_data.merge(total_sales_per_order,on='orderID',how='left')
```

```python
ml_data
```

```python
order_details
```

```python
unique_products_per_order =order_details.groupby('orderID')['productID'].nunique().reset_index(name="unique_products")
unique_products_per_order
ml_data=ml_data.merge(unique_products_per_order,on="orderID")
```

```python
ml_data
```

```python
ml_data.columns
```

```python
ml_data = ml_data.drop(columns=['orderID','orderDate','requiredDate','shippedDate','shipping_days'])
```

```python
ml_data.columns
```

```python
x = ml_data.drop(columns=['shipping status'])
y = ml_data['shipping status']
```

```python
x.shape
```

```python
y.value_counts()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, stratify=y, random_state=42)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
y_train.value_counts()
```

```python
y_test.value_counts()
```

```python
numeric_features = ['freight','year','month','day_of_week','allowed_days','total_quantity_per_order','sales','unique_products']

categorical_features = ['customerID', 'employeeID','shipperID']
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
preprocessor=ColumnTransformer([('num',StandardScaler(),numeric_features),('cat',OneHotEncoder(handle_unknown='ignore'),categorical_features)])
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
model=Pipeline([('preprocessor',preprocessor),('classifier',LogisticRegression(max_iter=1000))])
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
pd.Series(y_pred).value_counts()
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test,y_pred)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```

```python
model_balanced=Pipeline([('preprocessor',preprocessor),('classifier',LogisticRegression(max_iter=1000,class_weight='balanced'))])
model_balanced.fit(X_train,y_train)
```

```python
y_pred = model_balanced.predict(X_test)
pd.Series(y_pred).value_counts()
```

```python
confusion_matrix(y_test, y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

```python
from sklearn.model_selection import StratifiedKFold
cv=StratifiedKFold(n_splits=5,shuffle=True,random_state=42)
from sklearn.model_selection import cross_val_score
cv_scores=cross_val_score(model_balanced,X_train,y_train,cv=cv,scoring='f1')
```

```python
cv_scores.mean()
```

```python
cv_scores
```

```python
from sklearn.model_selection import GridSearchCV
param_grid={'classifier__C':[0.01, 0.1, 1, 10]}
grid=GridSearchCV(model_balanced,param_grid=param_grid,cv=cv,scoring='f1')
```

```python
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred_grid=grid.predict(X_test)
```

```python
y_pred_grid
```

```python
print(classification_report(y_test, y_pred_grid))
```

```python
from sklearn.tree import DecisionTreeClassifier
tree_model = Pipeline([('preprocessor', preprocessor),('classifier', DecisionTreeClassifier(class_weight='balanced',random_state=42))])
tree_scores = cross_val_score(tree_model,X_train,y_train,cv=cv,scoring='f1')
```

```python
tree_scores
```

```python
tree_scores.mean()
```

```python
from sklearn.ensemble import RandomForestClassifier
forest_model = Pipeline([('preprocessor', preprocessor),('classifier', RandomForestClassifier(n_estimators=100,class_weight='balanced',random_state=42))])
forest_scores = cross_val_score(forest_model,X_train,y_train,cv=cv,scoring='f1')
```
